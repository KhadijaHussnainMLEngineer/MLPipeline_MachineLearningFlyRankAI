## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
print("""

My selected lane is CTR / Engagement Opportunity Scoring.

Instead of doing broad, general research, this lane moves to a more specific
problem: looking at pages that are appearing in search results but not getting
the engagement they should be. The goal isn't to find WHY those pages get low
engagement, it's to identify WHICH pages those are. We find which pages are
ranking but underperforming, and flag them so the problem can be solved.
""")
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
print("""

DECISION:
My model reviews content pages and identifies which ones are ranking well in
search but not getting the engagement they should  this helps decide which
pages actually need editorial attention.

ACTION:
The model provides a ranked list to the person or client responsible for the
content. They open the list and go through the flagged pages, checking and
correcting things like the title, meta description, or other elements that
may be causing the low engagement.

COST OF A WRONG CALL:
- False positive: if my results wrongly flag a page as low-engagement when
  it isn't, the client wastes time working on a page that didn't need any
  changes. Worse, they might end up editing a page that already had a good
  title, description, and content  unnecessarily disturbing something that
  was working fine.

- False negative: if a real low-engagement page is missed, it keeps sitting
  there ranking well but not converting clicks, and nobody addresses it.
""")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/KhadijaHussnainMLEngineer/MLPipeline_MachineLearningFlyRankAI/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print("SECTION 3: QUICK LOOK AT THE DATA")
print("-----------------------------------")
print("Total pages in dataset:", df.shape[0])
print("Total columns:", df.shape[1])
print()

# Filter to "visible" pages: ranking in top 20, with real search volume
visible = df[(df["avg_position"] > 0) & (df["avg_position"] <= 20) & (df["impressions_90d"] >= 500)]
print("Pages ranking in top 20 with 500+ impressions (our candidate pool):", visible.shape[0])
print()

print("CTR stats for visible pages:")
print(visible["ctr"].describe())
print()

low_ctr_count = visible[visible["ctr"] < visible["ctr"].median()].shape[0]
print(f"Pages below median CTR among visible pool: {low_ctr_count} out of {visible.shape[0]}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
print("""

WHAT I CAN CLAIM:
- Observed patterns: which pages show a gap between their ranking position
  and their actual click-through rate, compared to similar pages.
- Directional signals: whether a page's engagement is unusually low or high
  relative to its peers in the same position tier.
- Decision-support: a ranked list to help a reviewer prioritize which pages
  to check first.

WHAT I CANNOT CLAIM:
- I cannot prove WHY a page has low CTR (title, meta, intent mismatch, or
  a SERP feature stealing clicks are all possible causes I cannot isolate).
- I cannot claim fixing a flagged page WILL increase its clicks.
- I cannot claim to predict or explain Google's ranking algorithm.
- I cannot claim causal proof of anything — only correlation/association
  from observed data.
""")
```
